```python
import numpy as np
def moving_average_predictor(data, window_size=3):
    predictions = []
    for i in range(len(data) - window_size):
        window = data[i:i+window_size]
        predictions.append(np.mean(window))
    return predictions
```

```python
import pandas as pd

# Load the saved workload data
workload_data = pd.read_csv('data/workload_simulation.csv')

# Quick check
print(workload_data.head())  # Shows the first 5 rows
print("Workload data loaded successfully ✅")
```

```python
# Prepare predictions
model_a_preds = moving_average_predictor(workload_data['Model_A_Traffic'])
model_b_preds = moving_average_predictor(workload_data['Model_B_Chatbot'])
model_c_preds = moving_average_predictor(workload_data['Model_C_Speech'])

# Actual next points
actual_a = workload_data['Model_A_Traffic'][3:].reset_index(drop=True)
actual_b = workload_data['Model_B_Chatbot'][3:].reset_index(drop=True)
actual_c = workload_data['Model_C_Speech'][3:].reset_index(drop=True)
```

```python
prediction_df = pd.DataFrame({
    'Hour': workload_data['Hour'][3:].reset_index(drop=True),
    'Model_A_Actual': actual_a,
    'Model_A_Predicted': model_a_preds,
    'Model_B_Actual': actual_b,
    'Model_B_Predicted': model_b_preds,
    'Model_C_Actual': actual_c,
    'Model_C_Predicted': model_c_preds
})

# Show result
print(prediction_df.head())
```

```python
import os

# Create folder if not exists
os.makedirs('../results/plots', exist_ok=True)
```
