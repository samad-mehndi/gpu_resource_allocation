```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Time range: 24 hours
hours = np.arange(0, 24)
```

```python
# Model A: Traffic Monitoring - steady usage
model_a = 45 + np.random.normal(0, 3, size=24)

# Model B: Chatbot - peaks during work hours
model_b = 30 + np.random.normal(0, 5, size=24)
model_b[9:13] += 40  # morning spike
model_b[17:21] += 30  # evening spike

# Model C: Speech Recognition - mild fluctuations
model_c = 40 + 10 * np.sin(np.linspace(0, 3*np.pi, 24)) + np.random.normal(0, 5, size=24)

# Clip values between 0 and 100
model_a = np.clip(model_a, 0, 100)
model_b = np.clip(model_b, 0, 100)
model_c = np.clip(model_c, 0, 100)
```

```python
# Create DataFrame
workload_data = pd.DataFrame({
    'Hour': hours,
    'Model_A_Traffic': model_a,
    'Model_B_Chatbot': model_b,
    'Model_C_Speech': model_c
})
```

```python
# Plot
plt.figure(figsize=(12, 6))
plt.plot(hours, model_a, label='Model A - Traffic Monitoring')
plt.plot(hours, model_b, label='Model B - Chatbot')
plt.plot(hours, model_c, label='Model C - Speech Recognition')
plt.xlabel('Hour of the Day')
plt.ylabel('GPU Demand (%)')
plt.title('Simulated GPU Workload over 24 Hours')
plt.legend()
plt.grid(True)
plt.show()

# Display the data
print(workload_data)
```

```python
import os

# Create data directory if it doesn't exist
os.makedirs('data', exist_ok=True)

# Now save the file
workload_data.to_csv('data/workload_simulation.csv', index=False)
print("Saved workload simulation successfully ✅")
```
